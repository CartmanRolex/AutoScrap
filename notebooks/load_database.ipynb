# Load AutoScrap Database

This notebook loads `data/autoscrap.db` into a pandas DataFrame.

```python
from pathlib import Path
import sqlite3

import pandas as pd
```

```python
PROJECT_ROOT = Path.cwd()

# If you open this notebook from the notebooks/ folder, move one level up.
if PROJECT_ROOT.name == "notebooks":
    PROJECT_ROOT = PROJECT_ROOT.parent

DB_PATH = PROJECT_ROOT / "data" / "autoscrap.db"
DB_PATH
```

```python
conn = sqlite3.connect(DB_PATH)

df = pd.read_sql_query(
    "SELECT * FROM listings ORDER BY first_seen_at DESC",
    conn,
)

df.head()
```
