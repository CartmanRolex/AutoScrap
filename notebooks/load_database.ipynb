# Load AutoScrap Database

This notebook loads `data/autoscrap.db` into a pandas DataFrame.

```python
from pathlib import Path
import sqlite3

import pandas as pd
```

```python
PROJECT_ROOT = Path.cwd()

# If you open this notebook from the notebooks/ folder, move one level up.
if PROJECT_ROOT.name == "notebooks":
    PROJECT_ROOT = PROJECT_ROOT.parent

DB_PATH = PROJECT_ROOT / "data" / "autoscrap.db"
DB_PATH
```

```python
conn = sqlite3.connect(DB_PATH)

# Full table, including cross-posted duplicates.
df_all = pd.read_sql_query(
    "SELECT * FROM listings ORDER BY first_seen_at DESC",
    conn,
)

# Preferred analysis dataset: one row per canonical car/listing.
# When the same car is cross-posted from AutoScout24 to Anibis,
# deduped_listings keeps the AutoScout24 row if it exists.
df = pd.read_sql_query(
    "SELECT * FROM deduped_listings ORDER BY first_seen_at DESC",
    conn,
)

df.head()
```
